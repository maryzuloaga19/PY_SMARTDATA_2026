```python
dbutils.widgets.removeAll()
```

```python
dbutils.widgets.text("storageName", "pysmartdata2026")
dbutils.widgets.text("catalogo", "catalog_pysmartdata")
```

```python
storageName = dbutils.widgets.get("storageName")
catalogo = dbutils.widgets.get("catalogo")

ruta_bronze = f"abfss://bronze@{storageName}.dfs.core.windows.net"
ruta_silver = f"abfss://silver@{storageName}.dfs.core.windows.net"
ruta_golden = f"abfss://golden@{storageName}.dfs.core.windows.net"
```

### Eliminacion tablas Bronze

```python
%sql
-- DROP TABLES
DROP TABLE IF EXISTS ${catalogo}.bronze.facturacion;
DROP TABLE IF EXISTS ${catalogo}.bronze.recaudos;
```

```python
# REMOVE DATA (Bronze)
dbutils.fs.rm(f"{ruta_bronze}/facturacion", True)
dbutils.fs.rm(f"{ruta_bronze}/recaudos", True)
```

### Eliminacion tablas Silver

```python
%sql
-- DROP TABLES
DROP TABLE IF EXISTS ${catalogo}.silver.facturacion_transformed;
DROP TABLE IF EXISTS ${catalogo}.silver.facturacion_rechazados;
DROP TABLE IF EXISTS ${catalogo}.silver.recaudos_transformed;
DROP TABLE IF EXISTS ${catalogo}.silver.recaudos_rechazados;
```

```python
# REMOVE DATA (Silver)
dbutils.fs.rm(f"{ruta_silver}/facturacion_transformed", True)
dbutils.fs.rm(f"{ruta_silver}/facturacion_rechazados", True)
dbutils.fs.rm(f"{ruta_silver}/recaudos_transformed", True)
dbutils.fs.rm(f"{ruta_silver}/recaudos_rechazados", True)
```

### Eliminacion tablas Golden

```python
%sql
-- DROP TABLES
DROP TABLE IF EXISTS ${catalogo}.golden.kpi_facturacion_periodo;
DROP TABLE IF EXISTS ${catalogo}.golden.kpi_recaudo_periodo;
DROP TABLE IF EXISTS ${catalogo}.golden.kpi_cartera_suscriptor;
DROP TABLE IF EXISTS ${catalogo}.golden.kpi_indice_recaudo;
DROP TABLE IF EXISTS ${catalogo}.golden.kpi_calidad_datos;
DROP TABLE IF EXISTS ${catalogo}.golden.kpi_detalle_suscriptor;
```

```python
# REMOVE DATA (Golden)
dbutils.fs.rm(f"{ruta_golden}/kpi_facturacion_periodo", True)
dbutils.fs.rm(f"{ruta_golden}/kpi_recaudo_periodo", True)
dbutils.fs.rm(f"{ruta_golden}/kpi_cartera_suscriptor", True)
dbutils.fs.rm(f"{ruta_golden}/kpi_indice_recaudo", True)
dbutils.fs.rm(f"{ruta_golden}/kpi_calidad_datos", True)
dbutils.fs.rm(f"{ruta_golden}/kpi_detalle_suscriptor", True)
```
